```python
# %% [code]
# ✅ Full FHE-aware CNN + Chebyshev head + Batched Encrypted Calibration
#    with comprehensive metrics (Acc, Precision, Recall, F1, AUROC)
#    for Identical Stronger Models on PathMNIST (A & B; B→A transfer pre/post)
#    + Per-sample "Before vs After" bars and per-step calibration loss curve.

# ------------------------------
# Imports
# ------------------------------
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Dense, Conv2D, MaxPooling2D, Dropout, BatchNormalization, Layer,
    GlobalAveragePooling2D, RandomFlip, RandomRotation, RandomTranslation, RandomZoom,
    SpatialDropout2D, GaussianNoise
)
from tensorflow.keras import regularizers
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping, Callback
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_recall_fscore_support,
    classification_report,
    roc_auc_score,
    confusion_matrix,
    accuracy_score,
)
from sklearn.utils.class_weight import compute_class_weight
import warnings

from medmnist import PathMNIST
from torchvision import transforms

# TenSEAL optional (guarded)
try:
    import tenseal as ts
    TENSEAL_AVAILABLE = True
except Exception:
    TENSEAL_AVAILABLE = False
    warnings.warn("TenSEAL not available; encrypted step will be simulated in plaintext.")

# ------------------------------
# Reproducibility
# ------------------------------
np.random.seed(42)
tf.random.set_seed(42)

# ------------------------------
# Chebyshev Softmax Layer
# ------------------------------
class ChebyshevSoftmax(Layer):
    def __init__(self, coeffs, **kwargs):
        super().__init__(**kwargs)
        self.coeffs = coeffs
    def call(self, inputs):
        # center + min-max to [-1,1] for Chebyshev stability
        x = inputs - tf.reduce_max(inputs, axis=1, keepdims=True)
        xmin = tf.reduce_min(x, axis=1, keepdims=True)
        xmax = tf.reduce_max(x, axis=1, keepdims=True)
        x_scaled = 2.0 * (x - xmin) / (xmax - xmin + 1e-8) - 1.0
        # T0..Tn
        T = [tf.ones_like(x_scaled), x_scaled]
        for _ in range(2, len(self.coeffs)):
            Tn = 2.0 * x_scaled * T[-1] - T[-2]
            T.append(Tn)
        poly = tf.add_n([c * Tn for c, Tn in zip(self.coeffs, T)])
        poly = tf.clip_by_value(poly, 1e-6, float("inf"))
        return poly / tf.reduce_sum(poly, axis=1, keepdims=True)

# ------------------------------
# NumPy Chebyshev helpers (for head-only eval & HE calibration)
# ------------------------------
def chebyshev_poly_softmax_numpy(logits, coeffs):
    x = logits - np.max(logits)
    x_scaled = 2.0 * (x - np.min(x)) / (np.max(x) - np.min(x) + 1e-12) - 1.0
    T = [np.ones_like(x_scaled), x_scaled.copy()]
    for _ in range(2, len(coeffs)):
        Tn = 2.0 * x_scaled * T[-1] - T[-2]
        T.append(Tn)
    poly = sum(c * Tn for c, Tn in zip(coeffs, T))
    poly = np.clip(poly, 1e-6, np.inf)
    return poly / np.sum(poly)

def poly_predict_numpy(features, W, b, coeffs):
    logits = np.dot(features, W) + b
    return chebyshev_poly_softmax_numpy(logits, coeffs)

def encrypted_predict(enc_feat, W, b, coeffs):
    if TENSEAL_AVAILABLE:
        logits = np.array([
            enc_feat.dot(W[:, j].tolist()).decrypt()[0] + b[j]
            for j in range(W.shape[1])
        ])
    else:
        logits = np.dot(enc_feat, W) + b
    return chebyshev_poly_softmax_numpy(logits, coeffs)

# ------------------------------
# Load PathMNIST (9 classes, RGB 28x28)
# ------------------------------
transform = transforms.Compose([transforms.ToTensor()])
train_dataset = PathMNIST(split='train', download=True, transform=transform)
test_dataset  = PathMNIST(split='test',  download=True, transform=transform)

X_train_full = train_dataset.imgs.astype(np.float32) / 255.0
y_train_full = train_dataset.labels.flatten().astype(np.int64)
X_test_full  = test_dataset.imgs.astype(np.float32) / 255.0
y_test_full  = test_dataset.labels.flatten().astype(np.int64)

X_train_full = X_train_full.reshape(-1, 28, 28, 3)
X_test_full  = X_test_full.reshape(-1, 28, 28, 3)
num_classes = 9

# ------------------------------
# Skewed Non-IID Partitions (A: 0–4, B: 5–8)
# ------------------------------
def skewed_data_partition(X, y, primary_labels, primary_ratio=0.8):
    Xp, yp = X[np.isin(y, primary_labels)], y[np.isin(y, primary_labels)]
    Xs, ys = X[~np.isin(y, primary_labels)], y[~np.isin(y, primary_labels)]
    sec_size = max(1, int((1.0 - primary_ratio) * len(Xp)))
    sec_size = min(sec_size, len(Xs))
    sel = np.random.choice(len(Xs), sec_size, replace=False)
    Xf, yf = np.concatenate([Xp, Xs[sel]]), np.concatenate([yp, ys[sel]])
    idx = np.arange(len(Xf)); np.random.shuffle(idx)
    return Xf[idx], yf[idx]

# Slightly more diversity at B to help its base reach ≥80%
X_A, y_A = skewed_data_partition(X_train_full, y_train_full, [0,1,2,3,4], primary_ratio=0.80)
X_B, y_B = skewed_data_partition(X_train_full, y_train_full, [5,6,7,8], primary_ratio=0.75)

# Stratified validation splits
X_A_tr, X_A_val, y_A_tr, y_A_val = train_test_split(X_A, y_A, test_size=0.2, stratify=y_A, random_state=42)
X_B_tr, X_B_val, y_B_tr, y_B_val = train_test_split(X_B, y_B, test_size=0.2, stratify=y_B, random_state=42)
y_A_tr_cat = to_categorical(y_A_tr, num_classes)
y_B_tr_cat = to_categorical(y_B_tr, num_classes)

# ------------------------------
# Class Weights (per-site, only on present classes)
# ------------------------------
def site_class_weights(y_train_site, n_classes):
    present = np.unique(y_train_site)
    weights = compute_class_weight(class_weight='balanced', classes=present, y=y_train_site)
    cw = {int(c): float(w) for c, w in zip(present, weights)}
    for c in range(n_classes):
        if c not in cw:
            cw[c] = 1.0
    return cw

cw_A = site_class_weights(y_A_tr, num_classes)
cw_B = site_class_weights(y_B_tr, num_classes)

# ------------------------------
# Stronger Identical Model (3 conv stages up to 256f) — 64-dim penultimate
# ------------------------------
cheb_coeffs = [0.5, 0.25, -0.12, 0.05, -0.01]

def conv_block(x, f, wd, p_drop):
    x = Conv2D(f, 3, padding="same", activation="relu",
               kernel_regularizer=regularizers.l2(wd))(x)
    x = BatchNormalization()(x)
    x = Conv2D(f, 3, padding="same", activation="relu",
               kernel_regularizer=regularizers.l2(wd))(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D()(x)
    x = SpatialDropout2D(p_drop)(x)
    return x

def build_strong_model(weight_decay=3e-4, dropout_mid=0.20, dropout_head=0.45):
    inp = Input(shape=(28,28,3))
    aug = tf.keras.Sequential([
        RandomFlip("horizontal"),
        RandomRotation(0.10),
        RandomTranslation(0.08, 0.08),
        RandomZoom(0.15),
    ], name="augmentation")
    x = aug(inp)
    x = GaussianNoise(0.05)(x)

    x = conv_block(x, 64,  weight_decay, dropout_mid)
    x = conv_block(x, 128, weight_decay, dropout_mid)
    x = conv_block(x, 256, weight_decay, dropout_mid)

    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation="relu", kernel_regularizer=regularizers.l2(weight_decay))(x)
    x = Dropout(dropout_head)(x)
    features = Dense(64, activation="relu", kernel_regularizer=regularizers.l2(weight_decay),
                     name="penultimate")(x)
    logits = Dense(num_classes, name="final_dense")(features)
    out = ChebyshevSoftmax(cheb_coeffs, name="cheb_softmax")(logits)
    return Model(inp, out), Model(inp, features)

# Build both models
model_A, feat_A = build_strong_model()
model_B, feat_B = build_strong_model()

# ------------------------------
# Optimizers, loss, callbacks
# ------------------------------
# Cross-entropy with *mild* label smoothing (stable & helps skew)
ce_loss = tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.08)

# Warmup + Cosine schedule whose LR can be *scaled* by a factor var
class WarmupCosineWithFactor(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr, steps_per_epoch, warmup_epochs=4,
                 first_decay_epochs=10, t_mul=1.5, m_mul=0.8, alpha=5e-5):
        super().__init__()
        self.base_lr   = tf.constant(base_lr, dtype=tf.float32)
        self.spe       = tf.constant(steps_per_epoch, dtype=tf.float32)
        self.warmup_steps = tf.cast(tf.math.maximum(1.0, self.spe * warmup_epochs), tf.float32)
        self.cosine = tf.keras.optimizers.schedules.CosineDecayRestarts(
            initial_learning_rate=self.base_lr,
            first_decay_steps=int(steps_per_epoch*first_decay_epochs),
            t_mul=t_mul, m_mul=m_mul, alpha=alpha
        )
        # multiplicative factor we can change from a callback
        self.factor = tf.Variable(1.0, trainable=False, dtype=tf.float32, name="lr_factor")
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        def warm():
            return self.base_lr * (step / self.warmup_steps)
        def cosine():
            return self.cosine(step - self.warmup_steps)
        return self.factor * tf.cond(step < self.warmup_steps, warm, cosine)
    def get_config(self):
        return {
            "base_lr": float(self.base_lr.numpy()),
            "steps_per_epoch": int(self.spe.numpy()),
            "warmup_steps": int(self.warmup_steps.numpy())
        }

# A ReduceLROnPlateau-like callback that scales schedule.factor
class ReduceOnPlateauFactor(Callback):
    def __init__(self, schedule, monitor="val_accuracy", factor=0.6, patience=4, min_lr=5e-5, verbose=1):
        super().__init__()
        self.schedule = schedule
        self.monitor  = monitor
        self.factor   = factor
        self.patience = patience
        self.min_lr   = min_lr
        self.verbose  = verbose
        self.best     = -np.inf
        self.wait     = 0
    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}
        current = logs.get(self.monitor)
        if current is None:
            return
        if current > self.best + 1e-6:
            self.best = current
            self.wait = 0
        else:
            self.wait += 1
            if self.wait >= self.patience:
                # compute current LR to enforce min_lr
                step = tf.cast(self.model.optimizer.iterations, tf.float32)
                curr_lr = float(self.schedule(step).numpy())
                new_factor = float(self.schedule.factor.numpy()) * self.factor
                if curr_lr > self.min_lr:
                    self.schedule.factor.assign(new_factor)
                    if self.verbose:
                        print(f"\nEpoch {epoch+1}: ReduceOnPlateauFactor scaling LR factor to {new_factor:.6f}")
                self.wait = 0

# Training config
total_epochs = 90
base_lr = 1e-3
batch_size = 128
steps_per_epoch_A = max(1, len(X_A_tr)//batch_size)
steps_per_epoch_B = max(1, len(X_B_tr)//batch_size)

lr_sched_A = WarmupCosineWithFactor(base_lr, steps_per_epoch_A, warmup_epochs=4)
lr_sched_B = WarmupCosineWithFactor(base_lr, steps_per_epoch_B, warmup_epochs=4)

optimizer_A = tf.keras.optimizers.Adam(learning_rate=lr_sched_A)
optimizer_B = tf.keras.optimizers.Adam(learning_rate=lr_sched_B)

callbacks_A = [
    ReduceOnPlateauFactor(lr_sched_A, monitor="val_accuracy", factor=0.6, patience=4, min_lr=5e-5, verbose=1),
    EarlyStopping(monitor="val_accuracy", patience=10, restore_best_weights=True, verbose=1),
]
callbacks_B = [
    ReduceOnPlateauFactor(lr_sched_B, monitor="val_accuracy", factor=0.6, patience=4, min_lr=5e-5, verbose=1),
    EarlyStopping(monitor="val_accuracy", patience=10, restore_best_weights=True, verbose=1),
]

# Eager to avoid Keras 3 graph numpy conversion quirks
model_A.compile(optimizer=optimizer_A, loss=ce_loss, metrics=["accuracy"], run_eagerly=True)
model_B.compile(optimizer=optimizer_B, loss=ce_loss, metrics=["accuracy"], run_eagerly=True)

# ------------------------------
# Train with class weights
# ------------------------------
hist_A = model_A.fit(
    X_A_tr, y_A_tr_cat,
    epochs=total_epochs, batch_size=batch_size,
    validation_data=(X_A_val, to_categorical(y_A_val, num_classes)),
    class_weight=cw_A,
    callbacks=callbacks_A, verbose=1
)
hist_B = model_B.fit(
    X_B_tr, y_B_tr_cat,
    epochs=total_epochs, batch_size=batch_size,
    validation_data=(X_B_val, to_categorical(y_B_val, num_classes)),
    class_weight=cw_B,
    callbacks=callbacks_B, verbose=1
)

# ------------------------------
# Metric utilities
# ------------------------------
def probs_and_preds(model, X):
    P = model.predict(X, verbose=0)
    y_pred = np.argmax(P, axis=1)
    return P, y_pred

def summarize_metrics(y_true_int, prob_mat, name=""):
    y_pred = np.argmax(prob_mat, axis=1)
    acc = accuracy_score(y_true_int, y_pred)
    prf_micro = precision_recall_fscore_support(y_true_int, y_pred, average="micro", zero_division=0)
    prf_macro = precision_recall_fscore_support(y_true_int, y_pred, average="macro", zero_division=0)
    prf_weighted = precision_recall_fscore_support(y_true_int, y_pred, average="weighted", zero_division=0)
    y_true_oh = to_categorical(y_true_int, prob_mat.shape[1])
    try:
        auc_ovr = roc_auc_score(y_true_oh, prob_mat, multi_class="ovr", average="macro")
    except Exception:
        auc_ovr = np.nan
    print(f"\n== {name} ==")
    print(f"Acc={acc:.4f}, AUROC={auc_ovr:.4f}")
    print(f"F1 micro/macro={prf_micro[2]:.4f}/{prf_macro[2]:.4f}")
    print(classification_report(y_true_int, y_pred, digits=3, zero_division=0))
    cm = confusion_matrix(y_true_int, y_pred)
    plt.figure(figsize=(5,4)); plt.imshow(cm, interpolation="nearest")
    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted"); plt.ylabel("True"); plt.colorbar(); plt.tight_layout(); plt.show()
    return {"acc": acc, "auc_macro_ovr": auc_ovr, "cm": cm,
            "prf_micro": prf_micro, "prf_macro": prf_macro, "prf_weighted": prf_weighted}

# ------------------------------
# Per-site evaluations (held-out validation) — Chebyshev head
# ------------------------------
P_A_val, _ = probs_and_preds(model_A, X_A_val)
_ = summarize_metrics(y_A_val, P_A_val, name="Institution A (Validation, Chebyshev)")

P_B_val, _ = probs_and_preds(model_B, X_B_val)
_ = summarize_metrics(y_B_val, P_B_val, name="Institution B (Validation, Chebyshev)")

# ------------------------------
# B→A Transfer (Pre-Calibration): B's features with A's head
# ------------------------------
final_dense_A = model_A.get_layer("final_dense")
W_A, b_A = final_dense_A.get_weights()  # (64,num_classes), (num_classes,)

def transfer_probs(extractor_B, W_head, b_head, X, batch_size=256):
    feats = extractor_B.predict(X, batch_size=batch_size, verbose=0)  # N x 64
    probs = np.zeros((feats.shape[0], b_head.shape[0]), dtype=np.float64)
    for i in range(feats.shape[0]):
        probs[i] = poly_predict_numpy(feats[i], W_head, b_head, cheb_coeffs)
    return probs

P_transfer_pre = transfer_probs(feat_B, W_A, b_A, X_test_full)
_ = summarize_metrics(y_test_full, P_transfer_pre, name="B→A Transfer (Pre-Calibration, Chebyshev)")

# ------------------------------
# Batched Encrypted Calibration (HE-friendly head-only updates)
#   - We also (a) track per-step loss across the K-sample batch
#     and (b) visualize BEFORE/AFTER on one concrete misclassified sample.
# ------------------------------
K = 128  # calibration batch
y_pred_transfer_pre = np.argmax(P_transfer_pre, axis=1)
mis_indices = np.where(y_pred_transfer_pre != y_test_full)[0]
if len(mis_indices) == 0:
    mis_indices = np.arange(min(K, len(y_test_full)))
sel_idx = mis_indices[:K]

# pick a specific misclassified sample to visualize (first from sel_idx)
vis_idx = int(sel_idx[0])

# Extract B features for selected samples
feats_sel = feat_B.predict(X_test_full[sel_idx], batch_size=256, verbose=0)  # (K, 64)
targets_sel = y_test_full[sel_idx]  # (K,)

# Optional HE: encrypt each feature vector
if TENSEAL_AVAILABLE:
    context = ts.context(ts.SCHEME_TYPE.CKKS, 8192, [60,40,40,60])
    context.global_scale = 2**40
    context.generate_galois_keys()
    enc_feats = [ts.ckks_vector(context, f.tolist()) for f in feats_sel]
else:
    enc_feats = feats_sel  # plaintext fallback

# Prepare one-hot targets (for plaintext reference)
true_onehots = np.zeros((len(sel_idx), b_A.shape[0]), dtype=np.float64)
true_onehots[np.arange(len(sel_idx)), targets_sel] = 1.0

# --- NEW: capture BEFORE for the visualized sample ---
feat_vis = feat_B.predict(X_test_full[vis_idx:vis_idx+1], verbose=0)[0]
scores_before = poly_predict_numpy(feat_vis, W_A, b_A, cheb_coeffs)
print("\n🔎 Misclassified Sample (Pre-Cal): index={}, True={}, Pred={}".format(
    vis_idx, int(y_test_full[vis_idx]), int(np.argmax(scores_before))))
for c, s in enumerate(scores_before):
    print(f"Class {c}: {s:.4f}")

# Batched head update (add/mul only, HE-friendly when using enc_feats)
lr = 0.02
weight_decay = 2e-4
steps = 12
loss_hist = []  # --- NEW: track loss across steps ---

for step in range(steps):
    grad_W = np.zeros_like(W_A)
    grad_b = np.zeros_like(b_A)
    step_loss = 0.0
    for i in range(len(sel_idx)):
        curr_prob = encrypted_predict(enc_feats[i], W_A, b_A, cheb_coeffs)  # length C
        err = true_onehots[i] - np.array(curr_prob)                         # length C
        step_loss += float(np.sum(err**2))
        feat_vec = feats_sel[i]
        grad_W += np.outer(feat_vec, err)
        grad_b += err
    # Apply update with L2 regularization
    W_A += lr * grad_W - weight_decay * W_A
    b_A += lr * grad_b - weight_decay * b_A
    loss_hist.append(step_loss / len(sel_idx))  # average L2 over batch (for smoother curve)

# --- NEW: capture AFTER for the same visualized sample & plot ---
scores_after = poly_predict_numpy(feat_vis, W_A, b_A, cheb_coeffs)
print("\n✅ Misclassified Sample (Post-Cal): index={}, True={}, Pred={}".format(
    vis_idx, int(y_test_full[vis_idx]), int(np.argmax(scores_after))))
for c, s in enumerate(scores_after):
    print(f"Class {c}: {s:.4f}")

# Visualization (bars + loss curve), matching your attached example
plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
x = np.arange(num_classes)
plt.bar(x-0.2, scores_before, width=0.4, label="Before")
plt.bar(x+0.2, scores_after,  width=0.4, label="After")
plt.xticks(x)
plt.title("Class Probabilities (Misclassified Sample)")
plt.legend()

plt.subplot(1,2,2)
plt.plot(loss_hist, marker='o')
plt.title("Calibration Loss Curve (Per-Step)")
plt.xlabel("Step")
plt.ylabel("L2 Loss")
plt.tight_layout()
plt.show()

# ------------------------------
# B→A Transfer (Post-Calibration)
# ------------------------------
P_transfer_post = transfer_probs(feat_B, W_A, b_A, X_test_full)
_ = summarize_metrics(y_test_full, P_transfer_post, name="B→A Transfer (Post-Calibration, Chebyshev)")

# ------------------------------
# Training curves
# ------------------------------
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(hist_A.history.get("accuracy",[]), label="A Train")
plt.plot(hist_A.history.get("val_accuracy",[]), label="A Val")
plt.title("Institution A Accuracy (PathMNIST)"); plt.legend()

plt.subplot(1,2,2)
plt.plot(hist_B.history.get("accuracy",[]), label="B Train")
plt.plot(hist_B.history.get("val_accuracy",[]), label="B Val")
plt.title("Institution B Accuracy (PathMNIST)"); plt.legend()
plt.tight_layout(); plt.show()

print("\nDone.")
```

```python
# %% [code]
# ✅ FHE-aware CNN + Chebyshev head + Batched Encrypted Calibration
#    with comprehensive metrics (Acc, Precision, Recall, F1, AUROC)
#    HETEROGENEOUS MODELS on PathMNIST (A & B; B→A transfer pre/post)
#    ── Update: push BOTH base validations ≥80% before calibration ──
#    Changes focused on site B only:
#      • Stronger heterogeneous CNN (Residual + Squeeze&Excitation + mild upsample to 56×56)
#      • Class-balanced oversampled training ds (+ tuned MixUp, lower label smoothing)
#      • Auto balanced fine-tune stage if B still < 0.80
#
#    Method (Chebyshev head + HE-friendly head calibration) remains unchanged.
#    ➕ Added: per-sample BEFORE/AFTER probability bars + per-step calibration loss curve.

%matplotlib inline
import os, sys, math
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Dense, Conv2D, SeparableConv2D, MaxPooling2D, Dropout, BatchNormalization, Layer,
    GlobalAveragePooling2D, RandomFlip, RandomRotation, RandomTranslation, RandomZoom,
    SpatialDropout2D, GaussianNoise, Add, Activation, Lambda, Reshape, Multiply
)
from tensorflow.keras import regularizers
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping, Callback, ModelCheckpoint
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_recall_fscore_support,
    classification_report,
    roc_auc_score,
    confusion_matrix,
    accuracy_score,
)
from sklearn.utils.class_weight import compute_class_weight
import warnings

# Optional HE (guarded)
try:
    import tenseal as ts
    TENSEAL_AVAILABLE = True
except Exception:
    TENSEAL_AVAILABLE = False
    warnings.warn("TenSEAL not available; encrypted step will be simulated in plaintext.")

# Stabilize TF (SpatialDropout sometimes triggers a harmless graph-layout warning)
try:
    tf.config.optimizer.set_jit(False)
except Exception:
    pass
try:
    tf.keras.mixed_precision.set_global_policy("float32")
except Exception:
    pass

# Reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# ------------------------------
# Chebyshev Softmax (HE-friendly)
# ------------------------------
class ChebyshevSoftmax(Layer):
    def __init__(self, coeffs, **kwargs):
        super().__init__(**kwargs)
        self.coeffs = coeffs
    def call(self, inputs):
        x = inputs - tf.reduce_max(inputs, axis=1, keepdims=True)
        xmin = tf.reduce_min(x, axis=1, keepdims=True)
        xmax = tf.reduce_max(x, axis=1, keepdims=True)
        x_scaled = 2.0 * (x - xmin) / (xmax - xmin + 1e-8) - 1.0
        T = [tf.ones_like(x_scaled), x_scaled]
        for _ in range(2, len(self.coeffs)):
            Tn = 2.0 * x_scaled * T[-1] - T[-2]
            T.append(Tn)
        poly = tf.add_n([c * Tn for c, Tn in zip(self.coeffs, T)])
        poly = tf.clip_by_value(poly, 1e-6, float("inf"))
        return poly / tf.reduce_sum(poly, axis=1, keepdims=True)

# ------------------------------
# NumPy Chebyshev helpers (head-only eval & HE calibration)
# ------------------------------
def chebyshev_poly_softmax_numpy(logits, coeffs):
    x = logits - np.max(logits)
    rng = (np.max(x) - np.min(x))
    x_scaled = 2.0 * (x - np.min(x)) / (rng + 1e-12) - 1.0
    T = [np.ones_like(x_scaled), x_scaled.copy()]
    for _ in range(2, len(coeffs)):
        Tn = 2.0 * x_scaled * T[-1] - T[-2]
        T.append(Tn)
    poly = sum(c * Tn for c, Tn in zip(coeffs, T))
    poly = np.clip(poly, 1e-6, np.inf)
    return poly / np.sum(poly)

def poly_predict_numpy(features, W, b, coeffs):
    logits = np.dot(features, W) + b
    return chebyshev_poly_softmax_numpy(logits, coeffs)

def encrypted_predict(enc_feat, W, b, coeffs):
    if TENSEAL_AVAILABLE:
        logits = np.array([
            enc_feat.dot(W[:, j].tolist()).decrypt()[0] + b[j]
            for j in range(W.shape[1])
        ])
    else:
        logits = np.dot(enc_feat, W) + b
    return chebyshev_poly_softmax_numpy(logits, coeffs)

# ------------------------------
# Data: PathMNIST (9 classes, RGB 28x28)
# ------------------------------
from medmnist import PathMNIST
from torchvision import transforms

transform = transforms.Compose([transforms.ToTensor()])
train_dataset = PathMNIST(split='train', download=True, transform=transform)
test_dataset  = PathMNIST(split='test',  download=True, transform=transform)

X_train_full = train_dataset.imgs.astype(np.float32) / 255.0
y_train_full = train_dataset.labels.flatten().astype(np.int64)
X_test_full  = test_dataset.imgs.astype(np.float32) / 255.0
y_test_full  = test_dataset.labels.flatten().astype(np.int64)

X_train_full = X_train_full.reshape(-1, 28, 28, 3)
X_test_full  = X_test_full.reshape(-1, 28, 28, 3)
num_classes = 9

# ------------------------------
# Skewed Non-IID Partitions
#   A: primary 0–4 (92% kept)
#   B: primary 5–8 (85% kept)
# ------------------------------
def skewed_data_partition(X, y, primary_labels, primary_ratio=0.92):
    Xp, yp = X[np.isin(y, primary_labels)], y[np.isin(y, primary_labels)]
    Xs, ys = X[~np.isin(y, primary_labels)], y[~np.isin(y, primary_labels)]
    sec_size = int((1.0 - primary_ratio) * len(Xp))
    sec_size = max(1, min(sec_size, len(Xs)))
    sel = np.random.choice(len(Xs), sec_size, replace=False)
    Xf, yf = np.concatenate([Xp, Xs[sel]]), np.concatenate([yp, ys[sel]])
    idx = np.arange(len(Xf)); np.random.shuffle(idx)
    return Xf[idx], yf[idx]

X_A, y_A = skewed_data_partition(X_train_full, y_train_full, [0,1,2,3,4], primary_ratio=0.92)
X_B, y_B = skewed_data_partition(X_train_full, y_train_full, [5,6,7,8], primary_ratio=0.85)

# Stratified validation splits
X_A_tr, X_A_val, y_A_tr, y_A_val = train_test_split(X_A, y_A, test_size=0.2, stratify=y_A, random_state=42)
X_B_tr, X_B_val, y_B_tr, y_B_val = train_test_split(X_B, y_B, test_size=0.2, stratify=y_B, random_state=42)

# ------------------------------
# Class Weights (per-site)
# ------------------------------
def site_class_weights(y_train_site, n_classes):
    present = np.unique(y_train_site)
    weights = compute_class_weight(class_weight='balanced', classes=present, y=y_train_site)
    cw = {int(c): float(w) for c, w in zip(present, weights)}
    for c in range(n_classes):
        if c not in cw:
            cw[c] = 1.0
    return cw

cw_A = site_class_weights(y_A_tr, num_classes)
cw_B = site_class_weights(y_B_tr, num_classes)

# ------------------------------
# Oversample to class-balanced dataset (used for B)
# ------------------------------
def make_class_balanced_arrays(X, y, n_classes):
    idx_by_class = [np.where(y == c)[0] for c in range(n_classes)]
    cls_present = [i for i, idx in enumerate(idx_by_class) if len(idx) > 0]
    if len(cls_present) == 0:
        return X, y
    max_count = max(len(idx_by_class[c]) for c in cls_present)
    new_idx = []
    rng = np.random.default_rng(42)
    for c in cls_present:
        k = len(idx_by_class[c])
        if k == 0:
            continue
        take = idx_by_class[c]
        if k < max_count:
            extra = rng.choice(take, size=max_count - k, replace=True)
            take = np.concatenate([take, extra])
        new_idx.append(take)
    new_idx = np.concatenate(new_idx)
    rng.shuffle(new_idx)
    return X[new_idx], y[new_idx]

# ------------------------------
# MixUp tf.data pipeline (supports sample weights)
# ------------------------------
def make_train_ds(X, y_int, class_weight_dict, batch_size, n_classes, mix_alpha=0.4, mix_prob=0.8, shuffle=True):
    sw = np.array([class_weight_dict[int(c)] for c in y_int], dtype=np.float32)
    y_oh = tf.one_hot(tf.convert_to_tensor(y_int, dtype=tf.int32), n_classes, dtype=tf.float32)

    ds = tf.data.Dataset.from_tensor_slices((X.astype(np.float32), y_oh, sw))
    if shuffle:
        ds = ds.shuffle(len(X), reshuffle_each_iteration=True)
    ds = ds.batch(batch_size)

    def beta_lam(alpha, shape):
        g1 = tf.random.gamma(shape, alpha, 1.0)
        g2 = tf.random.gamma(shape, alpha, 1.0)
        return g1 / (g1 + g2)

    def mixup(images, labels, sweights, alpha):
        bs = tf.shape(images)[0]
        idx = tf.random.shuffle(tf.range(bs))
        images2 = tf.gather(images, idx)
        labels2 = tf.gather(labels, idx)
        sw2     = tf.gather(sweights, idx)
        lam = beta_lam(alpha, [bs,1,1,1])
        lam_label = tf.reshape(lam, [bs,1])
        images_m = lam*images + (1.0-lam)*images2
        labels_m = lam_label*labels + (1.0-lam_label)*labels2
        sw_m     = lam[:,0,0,0]*sweights + (1.0-lam[:,0,0,0])*sw2
        return images_m, labels_m, sw_m

    def map_fn(images, labels, sweights):
        apply = tf.less(tf.random.uniform([], 0, 1), mix_prob)
        images_m, labels_m, sw_m = tf.cond(
            apply,
            lambda: mixup(images, labels, sweights, mix_alpha),
            lambda: (images, labels, sweights)
        )
        return images_m, labels_m, sw_m

    return ds.map(map_fn, num_parallel_calls=tf.data.AUTOTUNE).prefetch(tf.data.AUTOTUNE)

# ------------------------------
# Heterogeneous Models (A vs B) — both yield 64-dim penultimate
# ------------------------------
cheb_coeffs = [0.5, 0.25, -0.12, 0.05, -0.01]

def conv_block(x, f, wd, p_drop):
    x = Conv2D(f, 3, padding="same", activation="relu",
               kernel_regularizer=regularizers.l2(wd))(x)
    x = BatchNormalization()(x)
    x = Conv2D(f, 3, padding="same", activation="relu",
               kernel_regularizer=regularizers.l2(wd))(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D()(x)
    x = SpatialDropout2D(p_drop)(x)
    return x

# Simple SE (squeeze-excitation) block (Keras-3 safe)
def se_block(x, reduction=8):
    ch = int(x.shape[-1])
    s = GlobalAveragePooling2D()(x)
    s = Dense(max(ch // reduction, 4), activation="relu")(s)
    s = Dense(ch, activation="sigmoid")(s)
    s = Reshape((1, 1, ch))(s)
    return Multiply()([x, s])

# Residual Separable block (B only)
def res_sep_block(x, f, wd, stride=1, p_drop=0.10):
    shortcut = x
    x = SeparableConv2D(f, 3, padding="same", strides=stride,
                        depthwise_regularizer=regularizers.l2(wd),
                        pointwise_regularizer=regularizers.l2(wd))(x)
    x = BatchNormalization()(x); x = Activation("relu")(x)
    x = SeparableConv2D(f, 3, padding="same",
                        depthwise_regularizer=regularizers.l2(wd),
                        pointwise_regularizer=regularizers.l2(wd))(x)
    x = BatchNormalization()(x)
    if stride != 1 or int(shortcut.shape[-1]) != f:
        shortcut = Conv2D(f, 1, strides=stride, padding="same",
                          kernel_regularizer=regularizers.l2(wd))(shortcut)
        shortcut = BatchNormalization()(shortcut)
    x = Add()([x, shortcut])
    x = Activation("relu")(x)
    x = se_block(x)
    x = SpatialDropout2D(p_drop)(x)
    x = MaxPooling2D()(x)
    return x

# Institution A model — unchanged (already strong)
def build_model_A(weight_decay=2.5e-4, dropout_mid=0.12, dropout_head=0.30):
    inp = Input(shape=(28,28,3))
    aug = tf.keras.Sequential([
        RandomFlip("horizontal"),
        RandomRotation(0.06),
        RandomTranslation(0.05, 0.05),
        RandomZoom(0.10),
    ], name="augmentation_A")
    x = aug(inp)
    x = GaussianNoise(0.03)(x)
    x = conv_block(x, 64,  weight_decay, dropout_mid)
    x = conv_block(x, 128, weight_decay, dropout_mid)
    x = conv_block(x, 256, weight_decay, dropout_mid)
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation="relu", kernel_regularizer=regularizers.l2(weight_decay))(x)
    x = Dropout(dropout_head)(x)
    features = Dense(64, activation="relu", kernel_regularizer=regularizers.l2(weight_decay),
                     name="penultimate")(x)
    logits = Dense(num_classes, name="final_dense")(features)
    out = ChebyshevSoftmax(cheb_coeffs, name="cheb_softmax")(logits)
    return Model(inp, out), Model(inp, features)

# Institution B model — stronger & still heterogeneous
def build_model_B(weight_decay=2.0e-4, dropout_head=0.22):
    inp = Input(shape=(28,28,3))
    aug = tf.keras.Sequential([
        RandomFlip("horizontal"),
        RandomRotation(0.05),
        RandomTranslation(0.05, 0.05),
        RandomZoom(0.10),
    ], name="augmentation_B")
    x = aug(inp)
    # Mild upsample for larger receptive field on tiny inputs
    x = Lambda(lambda t: tf.image.resize(t, [56,56], method="bilinear"), name="upsample56")(x)
    # No extra GaussianNoise (B tended to underfit minority classes)
    x = res_sep_block(x,  96, weight_decay, stride=1, p_drop=0.08)
    x = res_sep_block(x, 192, weight_decay, stride=1, p_drop=0.10)
    x = res_sep_block(x, 256, weight_decay, stride=1, p_drop=0.12)
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation="relu", kernel_regularizer=regularizers.l2(weight_decay))(x)
    x = Dropout(dropout_head)(x)
    features = Dense(64, activation="relu", kernel_regularizer=regularizers.l2(weight_decay),
                     name="penultimate")(x)
    logits = Dense(num_classes, name="final_dense")(features)
    out = ChebyshevSoftmax(cheb_coeffs, name="cheb_softmax")(logits)
    return Model(inp, out), Model(inp, features)

# Build models
model_A, feat_A = build_model_A()
model_B, feat_B = build_model_B()

# ------------------------------
# Optimizers, loss, schedules, callbacks
# ------------------------------
ce_loss_A = tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.03)
ce_loss_B = tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.01)

class WarmupCosineWithFactor(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr, steps_per_epoch, warmup_epochs=3,
                 first_decay_epochs=22, t_mul=1.6, m_mul=0.9, alpha=8e-5):
        super().__init__()
        # store as plain Python floats/ints for serialization
        self._base_lr = float(base_lr)
        self._spe = int(steps_per_epoch)
        self._warmup_epochs = int(warmup_epochs)
        self._first_decay_epochs = int(first_decay_epochs)
        self._t_mul = float(t_mul)
        self._m_mul = float(m_mul)
        self._alpha = float(alpha)

        self.base_lr = tf.constant(self._base_lr, dtype=tf.float32)
        self.spe = tf.constant(self._spe, dtype=tf.float32)
        self.warmup_steps = tf.cast(tf.math.maximum(1.0, self.spe * self._warmup_epochs), tf.float32)
        self.cosine = tf.keras.optimizers.schedules.CosineDecayRestarts(
            initial_learning_rate=self.base_lr,
            first_decay_steps=int(self._spe*self._first_decay_epochs),
            t_mul=self._t_mul, m_mul=self._m_mul, alpha=self._alpha
        )
        # changed by ReduceOnPlateauFactor during training
        self.factor = tf.Variable(1.0, trainable=False, dtype=tf.float32, name="lr_factor")

    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        def warm():
            return self.base_lr * (step / self.warmup_steps)
        def cosine():
            return self.cosine(step - self.warmup_steps)
        return self.factor * tf.cond(step < self.warmup_steps, warm, cosine)

    def get_config(self):
        return {
            "base_lr": self._base_lr,
            "steps_per_epoch": self._spe,
            "warmup_epochs": self._warmup_epochs,
            "first_decay_epochs": self._first_decay_epochs,
            "t_mul": self._t_mul,
            "m_mul": self._m_mul,
            "alpha": self._alpha,
            "factor": float(self.factor.numpy()),
        }

    @classmethod
    def from_config(cls, cfg):
        obj = cls(
            base_lr=cfg["base_lr"],
            steps_per_epoch=cfg["steps_per_epoch"],
            warmup_epochs=cfg.get("warmup_epochs", 3),
            first_decay_epochs=cfg.get("first_decay_epochs", 22),
            t_mul=cfg.get("t_mul", 1.6),
            m_mul=cfg.get("m_mul", 0.9),
            alpha=cfg.get("alpha", 8e-5),
        )
        try:
            obj.factor.assign(float(cfg.get("factor", 1.0)))
        except Exception:
            pass
        return obj

class ReduceOnPlateauFactor(Callback):
    def __init__(self, schedule, monitor="val_accuracy", factor=0.6, patience=4, min_lr=7.5e-5, verbose=1):
        super().__init__()
        self.schedule = schedule
        self.monitor  = monitor
        self.factor   = factor
        self.patience = patience
        self.min_lr   = min_lr
        self.verbose  = verbose
        self.best     = -np.inf
        self.wait     = 0
    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}
        current = logs.get(self.monitor)
        if current is None: return
        if current > self.best + 1e-6:
            self.best = current; self.wait = 0
        else:
            self.wait += 1
            if self.wait >= self.patience:
                step = tf.cast(self.model.optimizer.iterations, tf.float32)
                curr_lr = float(self.schedule(step).numpy())
                new_factor = float(self.schedule.factor.numpy()) * self.factor
                if curr_lr > self.min_lr:
                    self.schedule.factor.assign(new_factor)
                    if self.verbose:
                        print(f"\nEpoch {epoch+1}: ReduceOnPlateauFactor scaling LR factor to {new_factor:.6f}")
                self.wait = 0

# AdamW if available
try:
    AdamW = tf.keras.optimizers.AdamW
    use_adamw = True
except AttributeError:
    use_adamw = False

total_epochs = 120
batch_size   = 256

# Training datasets
steps_per_epoch_A = max(1, len(X_A_tr)//batch_size)
lr_sched_A = WarmupCosineWithFactor(base_lr=2.0e-3, steps_per_epoch=steps_per_epoch_A, warmup_epochs=3)

# B: class-balanced oversampled arrays + tuned MixUp
X_B_bal, y_B_bal = make_class_balanced_arrays(X_B_tr, y_B_tr, num_classes)
steps_per_epoch_B = max(1, len(X_B_bal)//batch_size)
lr_sched_B = WarmupCosineWithFactor(base_lr=2.6e-3, steps_per_epoch=steps_per_epoch_B, warmup_epochs=3)

if use_adamw:
    optimizer_A = AdamW(learning_rate=lr_sched_A, weight_decay=3e-4, clipnorm=1.0)
    optimizer_B = AdamW(learning_rate=lr_sched_B, weight_decay=2.5e-4, clipnorm=1.0)
else:
    optimizer_A = tf.keras.optimizers.Adam(learning_rate=lr_sched_A, clipnorm=1.0)
    optimizer_B = tf.keras.optimizers.Adam(learning_rate=lr_sched_B, clipnorm=1.0)

callbacks_A = [
    ReduceOnPlateauFactor(lr_sched_A, monitor="val_accuracy", factor=0.6, patience=4, min_lr=7.5e-5, verbose=1),
    ModelCheckpoint("best_A.weights.h5", monitor="val_accuracy", save_best_only=True, save_weights_only=True, verbose=0),
    EarlyStopping(monitor="val_accuracy", patience=16, restore_best_weights=True, verbose=1),
]
callbacks_B = [
    ReduceOnPlateauFactor(lr_sched_B, monitor="val_accuracy", factor=0.6, patience=5, min_lr=8e-5, verbose=1),
    ModelCheckpoint("best_B.weights.h5", monitor="val_accuracy", save_best_only=True, save_weights_only=True, verbose=0),
    EarlyStopping(monitor="val_accuracy", patience=18, restore_best_weights=True, verbose=1),
]

model_A.compile(optimizer=optimizer_A, loss=ce_loss_A, metrics=["accuracy"])
model_B.compile(optimizer=optimizer_B, loss=ce_loss_B, metrics=["accuracy"])

train_ds_A = make_train_ds(X_A_tr, y_A_tr, cw_A, batch_size, num_classes, mix_alpha=0.40, mix_prob=0.75)
train_ds_B = make_train_ds(X_B_bal, y_B_bal, cw_B, batch_size, num_classes, mix_alpha=0.35, mix_prob=0.80)

# ------------------------------
# Train Stage 1
# ------------------------------
hist_A = model_A.fit(
    train_ds_A,
    epochs=total_epochs,
    validation_data=(X_A_val, to_categorical(y_A_val, num_classes)),
    callbacks=callbacks_A, verbose=1
)
hist_B = model_B.fit(
    train_ds_B,
    epochs=total_epochs,
    validation_data=(X_B_val, to_categorical(y_B_val, num_classes)),
    callbacks=callbacks_B, verbose=1
)

# ------------------------------
# If any site < 0.80 val-acc, auto fine-tune
# ------------------------------
def evaluate_val_acc(model, X_val, y_val):
    P = model.predict(X_val, verbose=0)
    y_pred = np.argmax(P, axis=1)
    return accuracy_score(y_val, y_pred)

val_acc_A = evaluate_val_acc(model_A, X_A_val, y_A_val)
val_acc_B = evaluate_val_acc(model_B, X_B_val, y_B_val)

def balanced_finetune_B(model_B, X_bal, y_bal, cw, X_val, y_val, n_classes,
                        base_lr=8e-4, epochs=22, patience=8, wd=2.0e-4):
    # Freeze lower blocks, train top block + head on balanced data
    for layer in model_B.layers:
        layer.trainable = False
    for layer in model_B.layers[-10:]:
        layer.trainable = True
    try:
        AdamW = tf.keras.optimizers.AdamW
        opt = AdamW(learning_rate=base_lr, weight_decay=wd, clipnorm=1.0)
    except AttributeError:
        opt = tf.keras.optimizers.Adam(learning_rate=base_lr, clipnorm=1.0)
    model_B.compile(optimizer=opt,
                    loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.01),
                    metrics=["accuracy"])
    ds = make_train_ds(X_bal, y_bal, cw, batch_size, n_classes, mix_alpha=0.30, mix_prob=0.75)
    cb = [EarlyStopping(monitor="val_accuracy", patience=patience, restore_best_weights=True, verbose=1)]
    model_B.fit(ds, epochs=epochs,
                validation_data=(X_val, to_categorical(y_val, n_classes)),
                callbacks=cb, verbose=1)

def full_finetune(model, X_tr, y_tr, cw, X_val, y_val, n_classes,
                  base_lr=7e-4, epochs=26, patience=8, wd=2e-4, mix_alpha=0.40, mix_prob=0.88):
    for layer in model.layers:
        layer.trainable = True
    try:
        AdamW = tf.keras.optimizers.AdamW
        opt = AdamW(learning_rate=base_lr, weight_decay=wd, clipnorm=1.0)
    except AttributeError:
        opt = tf.keras.optimizers.Adam(learning_rate=base_lr, clipnorm=1.0)
    model.compile(optimizer=opt,
                  loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.01),
                  metrics=["accuracy"])
    ds = make_train_ds(X_tr, y_tr, cw, batch_size, n_classes, mix_alpha=mix_alpha, mix_prob=mix_prob)
    cb = [EarlyStopping(monitor="val_accuracy", patience=patience, restore_best_weights=True, verbose=1)]
    model.fit(ds, epochs=epochs,
              validation_data=(X_val, to_categorical(y_val, n_classes)),
              callbacks=cb, verbose=1)

if val_acc_A < 0.80:
    full_finetune(model_A, X_A_tr, y_A_tr, cw_A, X_A_val, y_A_val, num_classes,
                  base_lr=6e-4, epochs=24, patience=8, wd=2e-4, mix_alpha=0.50, mix_prob=0.90)
    val_acc_A = evaluate_val_acc(model_A, X_A_val, y_A_val)

if val_acc_B < 0.80:
    print(f"\n[Balanced FT] B val_acc={val_acc_B:.4f} < 0.80 → balanced head/top fine-tune…")
    balanced_finetune_B(model_B, X_B_bal, y_B_bal, cw_B, X_B_val, y_B_val, num_classes)
    val_acc_B = evaluate_val_acc(model_B, X_B_val, y_B_val)

if val_acc_B < 0.80:
    print(f"\n[Full FT] B val_acc={val_acc_B:.4f} < 0.80 → full fine-tune…")
    full_finetune(model_B, X_B_bal, y_B_bal, cw_B, X_B_val, y_B_val, num_classes)
    val_acc_B = evaluate_val_acc(model_B, X_B_val, y_B_val)

print(f"\nFinal validation accuracies →  A: {val_acc_A:.4f}   B: {val_acc_B:.4f}")

# ------------------------------
# Metric utilities
# ------------------------------
def probs_and_preds(model, X):
    P = model.predict(X, verbose=0)
    y_pred = np.argmax(P, axis=1)
    return P, y_pred

def summarize_metrics(y_true_int, prob_mat, name=""):
    y_pred = np.argmax(prob_mat, axis=1)
    acc = accuracy_score(y_true_int, y_pred)
    prf_micro = precision_recall_fscore_support(y_true_int, y_pred, average="micro", zero_division=0)
    prf_macro = precision_recall_fscore_support(y_true_int, y_pred, average="macro", zero_division=0)
    prf_weighted = precision_recall_fscore_support(y_true_int, y_pred, average="weighted", zero_division=0)
    y_true_oh = to_categorical(y_true_int, prob_mat.shape[1])
    try:
        auc_ovr = roc_auc_score(y_true_oh, prob_mat, multi_class="ovr", average="macro")
    except Exception:
        auc_ovr = np.nan
    print(f"\n== {name} ==")
    print(f"Acc={acc:.4f}, AUROC={auc_ovr:.4f}")
    print(f"F1 micro/macro={prf_micro[2]:.4f}/{prf_macro[2]:.4f}")
    print(classification_report(y_true_int, y_pred, digits=3, zero_division=0))
    cm = confusion_matrix(y_true_int, y_pred)
    plt.figure(figsize=(5,4)); plt.imshow(cm, interpolation="nearest")
    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted"); plt.ylabel("True"); plt.colorbar(); plt.tight_layout(); plt.show()
    return {"acc": acc, "auc_macro_ovr": auc_ovr, "cm": cm,
            "prf_micro": prf_micro, "prf_macro": prf_macro, "prf_weighted": prf_weighted}

# ------------------------------
# Per-site evaluations (held-out validation) — Chebyshev head
# ------------------------------
P_A_val, _ = probs_and_preds(model_A, X_A_val)
_ = summarize_metrics(y_A_val, P_A_val, name="Institution A (Validation, Chebyshev)")

P_B_val, _ = probs_and_preds(model_B, X_B_val)
_ = summarize_metrics(y_B_val, P_B_val, name="Institution B (Validation, Chebyshev)")

# ------------------------------
# B→A Transfer (Pre-Calibration): B's features with A's head
# ------------------------------
final_dense_A = model_A.get_layer("final_dense")
W_A, b_A = final_dense_A.get_weights()

def transfer_probs(extractor_B, W_head, b_head, X, batch_size=256):
    feats = extractor_B.predict(X, batch_size=batch_size, verbose=0)  # N x 64
    probs = np.zeros((feats.shape[0], b_head.shape[0]), dtype=np.float64)
    for i in range(feats.shape[0]):
        probs[i] = poly_predict_numpy(feats[i], W_head, b_head, cheb_coeffs)
    return probs

P_transfer_pre = transfer_probs(feat_B, W_A, b_A, X_test_full)
_ = summarize_metrics(y_test_full, P_transfer_pre, name="B→A Transfer (Pre-Calibration, Chebyshev)")

# ------------------------------
# Batched Encrypted Calibration (HE-friendly head-only updates)
# ➕ plus: single-sample BEFORE/AFTER & per-step loss curve
# ------------------------------
K = 128
y_pred_transfer_pre = np.argmax(P_transfer_pre, axis=1)
mis_indices = np.where(y_pred_transfer_pre != y_test_full)[0]
if len(mis_indices) == 0:
    mis_indices = np.arange(min(K, len(y_test_full)))
sel_idx = mis_indices[:K]

# Extract B features for selected samples
feats_sel = feat_B.predict(X_test_full[sel_idx], batch_size=256, verbose=0)
targets_sel = y_test_full[sel_idx]

# Choose a representative misclassified sample (first in the batch)
single_idx_global = int(sel_idx[0])
single_feat = feats_sel[0]
single_true = int(targets_sel[0])

# BEFORE probs for the chosen sample
scores_before = poly_predict_numpy(single_feat, W_A, b_A, cheb_coeffs)
print(f"\n🔎 Misclassified Sample (Pre-Cal): index={single_idx_global}, True={single_true}, Pred={np.argmax(scores_before)}")
for c, s in enumerate(scores_before):
    print(f"Class {c}: {s:.4f}")

# Optional HE: encrypt each feature vector
if TENSEAL_AVAILABLE:
    context = ts.context(ts.SCHEME_TYPE.CKKS, 8192, [60,40,40,60])
    context.global_scale = 2**40
    context.generate_galois_keys()
    enc_feats = [ts.ckks_vector(context, f.tolist()) for f in feats_sel]
else:
    enc_feats = feats_sel

# One-hot targets for the calibration batch
true_onehots = np.zeros((len(sel_idx), b_A.shape[0]), dtype=np.float64)
true_onehots[np.arange(len(sel_idx)), targets_sel] = 1.0

# Calibration hyperparams
lr = 0.02
weight_decay = 2e-4
steps = 20  # more steps for a nicer loss curve

loss_hist = []  # track batch L2 loss per step

for step in range(steps):
    grad_W = np.zeros_like(W_A)
    grad_b = np.zeros_like(b_A)
    loss_sum = 0.0

    for i in range(len(sel_idx)):
        curr_prob = encrypted_predict(enc_feats[i], W_A, b_A, cheb_coeffs)
        err = true_onehots[i] - np.array(curr_prob)
        loss_sum += float(np.sum(err**2))
        feat_vec = feats_sel[i]
        grad_W += np.outer(feat_vec, err)
        grad_b += err

    # Record mean L2 loss before the update at this step
    loss_hist.append(loss_sum / len(sel_idx))

    # Apply update with L2 regularization
    W_A += lr * grad_W - weight_decay * W_A
    b_A += lr * grad_b - weight_decay * b_A

# AFTER probs for the chosen sample
scores_after = poly_predict_numpy(single_feat, W_A, b_A, cheb_coeffs)
print(f"\n✅ Misclassified Sample (Post-Cal): index={single_idx_global}, True={single_true}, Pred={np.argmax(scores_after)}")
for c, s in enumerate(scores_after):
    print(f"Class {c}: {s:.4f}")

# Plots: BEFORE vs AFTER bars (single sample) + per-step loss curve
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
x = np.arange(num_classes)
plt.bar(x-0.2, scores_before, width=0.4, label="Before")
plt.bar(x+0.2, scores_after,  width=0.4, label="After")
plt.xticks(x)
plt.title("Class Probabilities (Misclassified Sample)")
plt.legend()

plt.subplot(1,2,2)
plt.plot(loss_hist, marker='o')
plt.title("Calibration Loss Curve (Per-Step)")
plt.xlabel("Step"); plt.ylabel("L2 Loss")
plt.tight_layout(); plt.show()

# ------------------------------
# B→A Transfer (Post-Calibration)
# ------------------------------
P_transfer_post = transfer_probs(feat_B, W_A, b_A, X_test_full)
_ = summarize_metrics(y_test_full, P_transfer_post, name="B→A Transfer (Post-Calibration, Chebyshev)")

# ------------------------------
# Training curves (for reference)
# ------------------------------
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(hist_A.history.get("accuracy",[]), label="A Train")
plt.plot(hist_A.history.get("val_accuracy",[]), label="A Val")
plt.title("Institution A Accuracy (PathMNIST)"); plt.legend()

plt.subplot(1,2,2)
plt.plot(hist_B.history.get("accuracy",[]), label="B Train")
plt.plot(hist_B.history.get("val_accuracy",[]), label="B Val")
plt.title("Institution B Accuracy (PathMNIST)"); plt.legend()
plt.tight_layout(); plt.show()

print("\nDone.")
```
