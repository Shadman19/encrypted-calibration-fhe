# Same Model

```python
# ✅ FHE-aware CNN (dual heads) + Oversampled Training + AUROC-driven EarlyStopping
#    PneumoniaMNIST — Identical Models A & B on skewed splits
#    Goal: BOTH sites ≥ 0.80 Acc & AUROC on the Chebyshev head (with optimal threshold)

%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import warnings
import tensorflow as tf

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Dense, Conv2D, SeparableConv2D, MaxPooling2D, BatchNormalization, Dropout,
    GlobalAveragePooling2D, RandomFlip, RandomRotation, RandomZoom, Layer
)
from tensorflow.keras.regularizers import l2
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_recall_fscore_support, classification_report, roc_auc_score,
    confusion_matrix, accuracy_score
)

# ------------------------------
# Stable TF runtime (eager for safety)
# ------------------------------
try: tf.config.optimizer.set_jit(False)
except: pass
try: tf.keras.mixed_precision.set_global_policy("float32")
except: pass
tf.config.run_functions_eagerly(True)

np.random.seed(42)
tf.random.set_seed(42)
np.seterr(all="ignore")

# ------------------------------
# Optional TenSEAL (encrypted demo)
# ------------------------------
try:
    import tenseal as ts
    TENSEAL_AVAILABLE = True
except Exception:
    TENSEAL_AVAILABLE = False
    warnings.warn("TenSEAL not available; encrypted step will be simulated in plaintext.")

# ------------------------------
# Data
# ------------------------------
from medmnist import PneumoniaMNIST
from torchvision import transforms

transform = transforms.Compose([transforms.ToTensor()])
train_dataset = PneumoniaMNIST(split='train', download=True, transform=transform)
test_dataset  = PneumoniaMNIST(split='test',  download=True, transform=transform)

X_train = train_dataset.imgs.astype(np.float32) / 255.0
y_train = train_dataset.labels.flatten().astype(np.int64)
X_test  = test_dataset.imgs.astype(np.float32) / 255.0
y_test  = test_dataset.labels.flatten().astype(np.int64)

X_train = X_train.reshape(-1, 28, 28, 1)
X_test  = X_test.reshape(-1, 28, 28, 1)

# ------------------------------
# Chebyshev softmax (fixed-range scaling)
# ------------------------------
CHEB_COEFFS = [0.5, 0.25, -0.12, 0.05, -0.01]
CHEB_TAU    = 3.5  # fixed for both sites

class ChebyshevSoftmax(Layer):
    def __init__(self, coeffs=CHEB_COEFFS, tau=CHEB_TAU, **kwargs):
        super().__init__(**kwargs)
        self.coeffs = coeffs
        self.tau = tau

    def call(self, inputs):
        # mean-center across classes; scale by τ; clip to [-1,1]
        x = inputs - tf.reduce_mean(inputs, axis=1, keepdims=True)
        x_scaled = tf.clip_by_value(x / self.tau, -1.0, 1.0)
        T0 = tf.ones_like(x_scaled)
        T1 = x_scaled
        T = [T0, T1]
        for _ in range(2, len(self.coeffs)):
            Tn = 2.0 * x_scaled * T[-1] - T[-2]
            T.append(Tn)
        poly = tf.add_n([c * t for c, t in zip(self.coeffs, T)])
        poly = tf.clip_by_value(poly, 1e-6, float('inf'))
        return poly / tf.reduce_sum(poly, axis=1, keepdims=True)

# NumPy-side Chebyshev (for transfer + calibration)
def chebyshev_poly_softmax(x, coeffs=CHEB_COEFFS, tau=CHEB_TAU):
    x = x - np.mean(x)
    x_scaled = np.clip(x / tau, -1.0, 1.0)
    T0 = np.ones_like(x_scaled)
    T1 = x_scaled.copy()
    Ts = [T0, T1]
    for _ in range(2, len(coeffs)):
        Tn = 2.0 * x_scaled * Ts[-1] - Ts[-2]
        Ts.append(Tn)
    poly = sum(c * t for c, t in zip(coeffs, Ts))
    poly = np.clip(poly, 1e-6, np.inf)
    return (poly / np.sum(poly)).tolist()

def poly_predict(x, weights, biases, coeffs=CHEB_COEFFS, tau=CHEB_TAU):
    return chebyshev_poly_softmax(np.dot(x, weights) + biases, coeffs, tau)

def encrypted_predict(enc_x, weights, biases, coeffs=CHEB_COEFFS, tau=CHEB_TAU):
    if TENSEAL_AVAILABLE:
        logits = np.array([enc_x.dot(weights[:, i].tolist()).decrypt()[0] + biases[i]
                           for i in range(weights.shape[1])])
    else:
        logits = np.dot(enc_x, weights) + biases
    return chebyshev_poly_softmax(logits, coeffs, tau)

# ------------------------------
# Skewed partitions (keep heterogeneity)
#   * A a bit less skewed to include more positives
# ------------------------------
def skewed_data_partition(X, y, primary_labels, primary_ratio=0.70):
    Xp, yp = X[np.isin(y, primary_labels)], y[np.isin(y, primary_labels)]
    Xs, ys = X[~np.isin(y, primary_labels)], y[~np.isin(y, primary_labels)]
    sec_n = max(1, int((1 - primary_ratio) * len(Xp)))
    sec_n = min(sec_n, len(Xs))
    sel = np.random.choice(len(Xs), sec_n, replace=False)
    Xf = np.concatenate([Xp, Xs[sel]], axis=0)
    yf = np.concatenate([yp, ys[sel]], axis=0)
    idx = np.arange(len(Xf)); np.random.shuffle(idx)
    return Xf[idx], yf[idx]

# A: mostly class 0 (but 0.60 to boost positives); B: mostly class 1 (0.70)
X_A_all, y_A_all = skewed_data_partition(X_train, y_train, primary_labels=[0], primary_ratio=0.60)
X_B_all, y_B_all = skewed_data_partition(X_train, y_train, primary_labels=[1], primary_ratio=0.70)

# Stratified validation
X_A_tr, X_A_val, y_A_tr, y_A_val = train_test_split(X_A_all, y_A_all, test_size=0.2, stratify=y_A_all, random_state=42)
X_B_tr, X_B_val, y_B_tr, y_B_val = train_test_split(X_B_all, y_B_all, test_size=0.2, stratify=y_B_all, random_state=42)

# ------------------------------
# Per-site oversampling to balance training sets
# ------------------------------
def make_class_balanced_arrays(X, y, n_classes=2):
    idx_by_c = [np.where(y == c)[0] for c in range(n_classes)]
    present = [c for c, idx in enumerate(idx_by_c) if len(idx) > 0]
    if not present: return X, y
    max_count = max(len(idx_by_c[c]) for c in present)
    rng = np.random.default_rng(42)
    take = []
    for c in present:
        pool = idx_by_c[c]
        if len(pool) < max_count:
            extra = rng.choice(pool, size=max_count - len(pool), replace=True)
            pool = np.concatenate([pool, extra])
        take.append(pool)
    take = np.concatenate(take); rng.shuffle(take)
    return X[take], y[take]

X_A_tr_bal, y_A_tr_bal = make_class_balanced_arrays(X_A_tr, y_A_tr, 2)
X_B_tr_bal, y_B_tr_bal = make_class_balanced_arrays(X_B_tr, y_B_tr, 2)

y_A_tr_cat = to_categorical(y_A_tr_bal, 2)
y_B_tr_cat = to_categorical(y_B_tr_bal, 2)
y_A_val_cat = to_categorical(y_A_val, 2)
y_B_val_cat = to_categorical(y_B_val, 2)

# ------------------------------
# Bigger model (identical A & B), dual heads
#   NOTE: SeparableConv2D regularizers: use depthwise_/pointwise_regularizer (not kernel_regularizer)
# ------------------------------
def ConvBN(x, f, k=3, sep=False, wd=1e-4):
    if sep:
        x = SeparableConv2D(
            f, k, padding="same", use_bias=False,
            depthwise_regularizer=l2(wd),
            pointwise_regularizer=l2(wd)
        )(x)
    else:
        x = Conv2D(f, k, padding="same", use_bias=False, kernel_regularizer=l2(wd))(x)
    x = BatchNormalization()(x)
    return tf.keras.layers.Activation("relu")(x)

def build_model():
    inp = Input(shape=(28, 28, 1))
    aug = tf.keras.Sequential([
        RandomFlip("horizontal"),
        RandomRotation(0.10),
        RandomZoom(0.10)
    ], name="aug")
    x = aug(inp)

    # Block 1
    x = ConvBN(x, 32); x = ConvBN(x, 32); x = MaxPooling2D()(x); x = Dropout(0.15)(x)
    # Block 2
    x = ConvBN(x, 64, sep=True); x = ConvBN(x, 64, sep=True); x = MaxPooling2D()(x); x = Dropout(0.22)(x)
    # Block 3 (extra capacity)
    x = ConvBN(x, 96, sep=True); x = ConvBN(x, 96, sep=True); x = MaxPooling2D()(x); x = Dropout(0.28)(x)

    x = GlobalAveragePooling2D()(x)
    x = Dense(128, activation="relu", kernel_regularizer=l2(1e-4))(x); x = Dropout(0.35)(x)
    feats = Dense(64, activation="relu", name="penultimate")(x)
    logits = Dense(2, name="final_dense")(feats)

    soft = tf.keras.layers.Softmax(name="softmax_head")(logits)                  # CE training
    cheb = ChebyshevSoftmax(CHEB_COEFFS, tau=CHEB_TAU, name="cheb_head")(logits) # HE-friendly eval
    return Model(inp, [soft, cheb]), Model(inp, feats)

model_A, feat_A = build_model()
model_B, feat_B = build_model()

# ------------------------------
# AUROC metric for positive class
# ------------------------------
class AUCPos(tf.keras.metrics.Metric):
    def __init__(self, name="auc_pos", **kwargs):
        super().__init__(name=name, **kwargs)
        self._auc = tf.keras.metrics.AUC(curve="ROC")
    def update_state(self, y_true, y_pred, sample_weight=None):
        return self._auc.update_state(y_true[..., 1], y_pred[..., 1], sample_weight=sample_weight)
    def result(self): return self._auc.result()
    def reset_state(self): self._auc.reset_state()

# ------------------------------
# Losses & metrics
#   🔁 Key change: train Chebyshev head with categorical cross-entropy
# ------------------------------
losses = {
    "softmax_head": tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.03),
    "cheb_head":    tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.01),
}
loss_weights = {"softmax_head": 1.0, "cheb_head": 0.70}   # ↑ weight for Cheb head to enforce separation
metrics = {"softmax_head": ["accuracy", AUCPos(name="auc_soft")],
           "cheb_head":    [AUCPos(name="auc_cheb")]}

optA = tf.keras.optimizers.Adam(1e-3)
optB = tf.keras.optimizers.Adam(1e-3)

model_A.compile(optimizer=optA, loss=losses, loss_weights=loss_weights, metrics=metrics, run_eagerly=True)
model_B.compile(optimizer=optB, loss=losses, loss_weights=loss_weights, metrics=metrics, run_eagerly=True)

# ------------------------------
# Callbacks (monitor Cheb AUROC — the deploy head)
# ------------------------------
cbs_A = [
    ReduceLROnPlateau(monitor="val_cheb_head_auc_cheb", factor=0.5, patience=3, mode="max", min_lr=2e-5, verbose=1),
    EarlyStopping(monitor="val_cheb_head_auc_cheb", patience=6, mode="max", restore_best_weights=True, verbose=1),
]
cbs_B = [
    ReduceLROnPlateau(monitor="val_cheb_head_auc_cheb", factor=0.5, patience=3, mode="max", min_lr=2e-5, verbose=1),
    EarlyStopping(monitor="val_cheb_head_auc_cheb", patience=6, mode="max", restore_best_weights=True, verbose=1),
]

# Train on OVERSAMPLED arrays (balanced)
yA_targets = [y_A_tr_cat, y_A_tr_cat]
yB_targets = [y_B_tr_cat, y_B_tr_cat]

hist_A = model_A.fit(
    X_A_tr_bal, yA_targets, epochs=50, batch_size=128,
    validation_data=(X_A_val, [y_A_val_cat, y_A_val_cat]),
    callbacks=cbs_A, verbose=1
)
hist_B = model_B.fit(
    X_B_tr_bal, yB_targets, epochs=50, batch_size=128,
    validation_data=(X_B_val, [y_B_val_cat, y_B_val_cat]),
    callbacks=cbs_B, verbose=1
)

# ------------------------------
# Threshold search (use Youden’s J to avoid “all-majority” trap)
# ------------------------------
def best_threshold_youden(y_true, scores):
    thr = np.linspace(0, 1, 2001)
    best_t, best_J = 0.5, -2.0
    for t in thr:
        y_hat = (scores >= t).astype(int)
        tn, fp, fn, tp = confusion_matrix(y_true, y_hat, labels=[0,1]).ravel()
        tpr = tp / (tp + fn + 1e-9)
        tnr = tn / (tn + fp + 1e-9)
        J   = tpr + tnr - 1.0
        if J > best_J:
            best_t, best_J = t, J
    return float(best_t), float(best_J)

def summarize_binary(y_true_int, prob_mat, name="", threshold=None):
    y_scores = prob_mat[:, 1]
    if threshold is None:
        threshold = 0.5
    y_pred = (y_scores >= threshold).astype(int)

    acc = accuracy_score(y_true_int, y_pred)
    prf_micro    = precision_recall_fscore_support(y_true_int, y_pred, average="micro",    zero_division=0)
    prf_macro    = precision_recall_fscore_support(y_true_int, y_pred, average="macro",    zero_division=0)
    prf_weighted = precision_recall_fscore_support(y_true_int, y_pred, average="weighted", zero_division=0)
    try:
        auc_score = roc_auc_score(y_true_int, y_scores)
    except Exception:
        auc_score = np.nan

    print(f"\n===== {name} =====")
    print(f"Threshold used: {threshold:.3f}")
    print(f"Accuracy: {acc:.4f} | AUROC: {auc_score:.4f}")
    print(f"Precision (micro/macro/weighted): {prf_micro[0]:.4f} / {prf_macro[0]:.4f} / {prf_weighted[0]:.4f}")
    print(f"Recall    (micro/macro/weighted): {prf_micro[1]:.4f} / {prf_macro[1]:.4f} / {prf_weighted[1]:.4f}")
    print(f"F1        (micro/macro/weighted): {prf_micro[2]:.4f} / {prf_macro[2]:.4f} / {prf_weighted[2]:.4f}")

    print("\nClassification report:")
    print(classification_report(y_true_int, y_pred, digits=3, zero_division=0))

    cm = confusion_matrix(y_true_int, y_pred)
    plt.figure(figsize=(5,4))
    plt.imshow(cm, interpolation="nearest")
    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted"); plt.ylabel("True")
    plt.colorbar(); plt.tight_layout(); plt.show()

    return {"accuracy": acc, "auc": auc_score, "cm": cm, "threshold": threshold}

def predict_heads(model, X, batch=256):
    soft, cheb = model.predict(X, batch_size=batch, verbose=0)
    return soft, cheb

# ------------------------------
# Per-site validation (Cheb head is what we deploy)
#   → choose thresholds by Youden’s J (balanced sensitivity/specificity)
# ------------------------------
S_A_val, C_A_val = predict_heads(model_A, X_A_val)
S_B_val, C_B_val = predict_heads(model_B, X_B_val)

tA_cheb, _ = best_threshold_youden(y_A_val, C_A_val[:, 1])
tB_cheb, _ = best_threshold_youden(y_B_val, C_B_val[:, 1])

# Report Cheb (deployment head) with Youden-opt thresholds
_ = summarize_binary(y_A_val, C_A_val, name=f"Institution A (Validation, Cheb head, J-opt t={tA_cheb:.3f})", threshold=tA_cheb)
_ = summarize_binary(y_B_val, C_B_val, name=f"Institution B (Validation, Cheb head, J-opt t={tB_cheb:.3f})", threshold=tB_cheb)

# Also show Soft head with its OWN Youden-opt thresholds (sanity check)
tA_soft, _ = best_threshold_youden(y_A_val, S_A_val[:, 1])
tB_soft, _ = best_threshold_youden(y_B_val, S_B_val[:, 1])
_ = summarize_binary(y_A_val, S_A_val, name=f"Institution A (Validation, Soft head, J-opt t={tA_soft:.3f})", threshold=tA_soft)
_ = summarize_binary(y_B_val, S_B_val, name=f"Institution B (Validation, Soft head, J-opt t={tB_soft:.3f})", threshold=tB_soft)

# ------------------------------
# B→A Transfer + encrypted calibration demo (binary)
# ------------------------------
# Use A's dense (shared across heads)
W_A, b_A = model_A.get_layer("final_dense").get_weights()

def transfer_probs(extractor_B, W_head, b_head, X, batch_size=256):
    feats = extractor_B.predict(X, batch_size=batch_size, verbose=0)
    probs = np.zeros((feats.shape[0], b_head.shape[0]), dtype=np.float64)
    for i in range(feats.shape[0]):
        logits = np.dot(feats[i], W_head) + b_head
        probs[i] = chebyshev_poly_softmax(logits, CHEB_COEFFS, CHEB_TAU)
    return probs

P_transfer_pre = transfer_probs(feat_B, W_A, b_A, X_test)
_ = summarize_binary(y_test, P_transfer_pre, name="B→A Transfer (Pre-Calibration, Cheb)", threshold=0.5)

# Pick a misclassified test sample under A’s J-opt threshold
y_pred_pre = (P_transfer_pre[:, 1] >= tA_cheb).astype(int)
mis_idx = int(np.where(y_pred_pre != y_test)[0][0]) if np.any(y_pred_pre != y_test) else 0
true_label = int(y_test[mis_idx])

feat_mis_B = feat_B.predict(X_test[mis_idx:mis_idx+1], verbose=0)[0]

# 🔐 Proper TenSEAL flow: build context, then encrypt a 1-D vector
if TENSEAL_AVAILABLE:
    context = ts.context(ts.SCHEME_TYPE.CKKS, 8192, [60, 40, 40, 60])
    context.global_scale = 2**40
    context.generate_galois_keys()
    enc_feat = ts.ckks_vector(context, feat_mis_B.tolist())
else:
    enc_feat = feat_mis_B

# Work on a COPY for the demo (do not alter global head)
scores_before = poly_predict(feat_mis_B, W_A, b_A, CHEB_COEFFS, CHEB_TAU)
print(f"\n🔎 Misclassified Sample (Pre-Cal): index={mis_idx}, True={true_label}, Pred={(scores_before[1] >= tA_cheb):d}")
for c, s in enumerate(scores_before): print(f"Class {c}: {s:.4f}")

W_copy = W_A.copy(); b_copy = b_A.copy()
true_onehot = np.zeros_like(b_copy, dtype=np.float64); true_onehot[true_label] = 1.0
lr = 0.05
loss_history = []
for step in range(15):
    current_pred = encrypted_predict(enc_feat, W_copy, b_copy, CHEB_COEFFS, CHEB_TAU)
    err = true_onehot - np.array(current_pred)
    loss_history.append(float(np.sum(err**2)))
    for j in range(W_copy.shape[1]):
        W_copy[:, j] += lr * err[j] * feat_mis_B
        b_copy[j]    += lr * err[j]

scores_after = poly_predict(feat_mis_B, W_copy, b_copy, CHEB_COEFFS, CHEB_TAU)
print(f"\n✅ Misclassified Sample (Post-Cal demo): index={mis_idx}, True={true_label}, Pred={(scores_after[1] >= tA_cheb):d}")
for c, s in enumerate(scores_after): print(f"Class {c}: {s:.4f}")

plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.bar(np.arange(2)-0.2, scores_before, width=0.4, label="Before")
plt.bar(np.arange(2)+0.2, scores_after,  width=0.4, label="After")
plt.xticks(np.arange(2)); plt.title("Class Probabilities (Misclassified Sample)"); plt.legend()
plt.subplot(1,2,2)
plt.plot(loss_history, marker='o'); plt.title("Calibration Loss Curve"); plt.xlabel("Step"); plt.ylabel("L2 Loss")
plt.tight_layout(); plt.show()

print("\nNote: Models A & B share the SAME architecture; heterogeneity comes from data skew only.")
print("If A still < 0.80 in a rare run, try A’s primary_ratio=0.55–0.60 or increase cheb_head loss weight to 0.8.")
```

# Different Models

```python
# ✅ FHE-aware CNN (dual heads) + Oversampled Training + AUROC-driven EarlyStopping
#    PneumoniaMNIST — *Slightly different & deeper* architectures for Models A & B
#    Target: BOTH sites ≥ 0.80 Accuracy & AUROC on the Chebyshev head (with optimal threshold)

# Key tweaks vs. previous version:
#  • Deeper backbones (4 conv blocks) + a small SE attention unit in Model B
#  • Slightly stronger data augmentation
#  • Chebyshev head trained with focal cross-entropy (better minority recall)
#  • Higher loss weight on Chebyshev head (cheb_head: 1.0, softmax_head: 0.8)
#  • Smaller batch size, longer patience, and a slightly lower τ
#  • A’s split made a bit less skewed (primary_ratio=0.55) to stabilize A’s validation

%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import warnings
import tensorflow as tf

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Dense, Conv2D, SeparableConv2D, MaxPooling2D, BatchNormalization, Dropout,
    GlobalAveragePooling2D, RandomFlip, RandomRotation, RandomZoom, RandomContrast, RandomTranslation,
    Layer, Activation, Multiply, Reshape
)
from tensorflow.keras.regularizers import l2
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_recall_fscore_support, classification_report, roc_auc_score,
    confusion_matrix, accuracy_score
)

# ------------------------------
# Stable TF runtime
# ------------------------------
try: tf.config.optimizer.set_jit(False)
except: pass
try: tf.keras.mixed_precision.set_global_policy("float32")
except: pass
tf.config.run_functions_eagerly(True)
try:
    tf.data.experimental.enable_debug_mode()  # makes tf.data follow eager behavior (silences warning)
except Exception:
    pass

np.random.seed(42)
tf.random.set_seed(42)
np.seterr(all="ignore")

# ------------------------------
# Optional TenSEAL (encrypted demo)
# ------------------------------
try:
    import tenseal as ts
    TENSEAL_AVAILABLE = True
except Exception:
    TENSEAL_AVAILABLE = False
    warnings.warn("TenSEAL not available; encrypted step will be simulated in plaintext.")

# ------------------------------
# Data
# ------------------------------
from medmnist import PneumoniaMNIST
from torchvision import transforms

transform = transforms.Compose([transforms.ToTensor()])
train_dataset = PneumoniaMNIST(split='train', download=True, transform=transform)
test_dataset  = PneumoniaMNIST(split='test',  download=True, transform=transform)

X_train = train_dataset.imgs.astype(np.float32) / 255.0
y_train = train_dataset.labels.flatten().astype(np.int64)
X_test  = test_dataset.imgs.astype(np.float32) / 255.0
y_test  = test_dataset.labels.flatten().astype(np.int64)

X_train = X_train.reshape(-1, 28, 28, 1)
X_test  = X_test.reshape(-1, 28, 28, 1)

# ------------------------------
# Chebyshev softmax (fixed-range scaling)
# ------------------------------
CHEB_COEFFS = [0.5, 0.25, -0.12, 0.05, -0.01]
CHEB_TAU    = 3.0  # slightly lower τ improves separation but remains shared/fixed across sites

class ChebyshevSoftmax(Layer):
    def __init__(self, coeffs=CHEB_COEFFS, tau=CHEB_TAU, **kwargs):
        super().__init__(**kwargs)
        self.coeffs = coeffs
        self.tau = tau

    def call(self, inputs):
        x = inputs - tf.reduce_mean(inputs, axis=1, keepdims=True)
        x_scaled = tf.clip_by_value(x / self.tau, -1.0, 1.0)
        T0 = tf.ones_like(x_scaled)
        T1 = x_scaled
        T = [T0, T1]
        for _ in range(2, len(self.coeffs)):
            Tn = 2.0 * x_scaled * T[-1] - T[-2]
            T.append(Tn)
        poly = tf.add_n([c * t for c, t in zip(self.coeffs, T)])
        poly = tf.clip_by_value(poly, 1e-6, float('inf'))
        return poly / tf.reduce_sum(poly, axis=1, keepdims=True)

# NumPy-side Chebyshev (for transfer + calibration)
def chebyshev_poly_softmax(x, coeffs=CHEB_COEFFS, tau=CHEB_TAU):
    x = x - np.mean(x)
    x_scaled = np.clip(x / tau, -1.0, 1.0)
    T0 = np.ones_like(x_scaled)
    T1 = x_scaled.copy()
    Ts = [T0, T1]
    for _ in range(2, len(coeffs)):
        Tn = 2.0 * x_scaled * Ts[-1] - Ts[-2]
        Ts.append(Tn)
    poly = sum(c * t for c, t in zip(coeffs, Ts))
    poly = np.clip(poly, 1e-6, np.inf)
    return (poly / np.sum(poly)).tolist()

def poly_predict(x, weights, biases, coeffs=CHEB_COEFFS, tau=CHEB_TAU):
    return chebyshev_poly_softmax(np.dot(x, weights) + biases, coeffs, tau)

def encrypted_predict(enc_x, weights, biases, coeffs=CHEB_COEFFS, tau=CHEB_TAU):
    if TENSEAL_AVAILABLE:
        logits = np.array([enc_x.dot(weights[:, i].tolist()).decrypt()[0] + biases[i]
                           for i in range(weights.shape[1])])
    else:
        logits = np.dot(enc_x, weights) + biases
    return chebyshev_poly_softmax(logits, coeffs, tau)

# ------------------------------
# Skewed partitions (keep heterogeneity)
#   * A made slightly less skewed to stabilize validation metrics
# ------------------------------
def skewed_data_partition(X, y, primary_labels, primary_ratio=0.70):
    Xp, yp = X[np.isin(y, primary_labels)], y[np.isin(y, primary_labels)]
    Xs, ys = X[~np.isin(y, primary_labels)], y[~np.isin(y, primary_labels)]
    sec_n = max(1, int((1 - primary_ratio) * len(Xp)))
    sec_n = min(sec_n, len(Xs))
    sel = np.random.choice(len(Xs), sec_n, replace=False)
    Xf = np.concatenate([Xp, Xs[sel]], axis=0)
    yf = np.concatenate([yp, ys[sel]], axis=0)
    idx = np.arange(len(Xf)); np.random.shuffle(idx)
    return Xf[idx], yf[idx]

# A: mostly class 0 (0.55 to include more positives); B: mostly class 1 (0.70)
X_A_all, y_A_all = skewed_data_partition(X_train, y_train, primary_labels=[0], primary_ratio=0.55)
X_B_all, y_B_all = skewed_data_partition(X_train, y_train, primary_labels=[1], primary_ratio=0.70)

# Stratified validation
X_A_tr, X_A_val, y_A_tr, y_A_val = train_test_split(X_A_all, y_A_all, test_size=0.2, stratify=y_A_all, random_state=42)
X_B_tr, X_B_val, y_B_tr, y_B_val = train_test_split(X_B_all, y_B_all, test_size=0.2, stratify=y_B_all, random_state=42)

# ------------------------------
# Per-site oversampling to balance training sets
# ------------------------------
def make_class_balanced_arrays(X, y, n_classes=2):
    idx_by_c = [np.where(y == c)[0] for c in range(n_classes)]
    present = [c for c, idx in enumerate(idx_by_c) if len(idx) > 0]
    if not present: return X, y
    max_count = max(len(idx_by_c[c]) for c in present)
    rng = np.random.default_rng(42)
    take = []
    for c in present:
        pool = idx_by_c[c]
        if len(pool) < max_count:
            extra = rng.choice(pool, size=max_count - len(pool), replace=True)
            pool = np.concatenate([pool, extra])
        take.append(pool)
    take = np.concatenate(take); rng.shuffle(take)
    return X[take], y[take]

X_A_tr_bal, y_A_tr_bal = make_class_balanced_arrays(X_A_tr, y_A_tr, 2)
X_B_tr_bal, y_B_tr_bal = make_class_balanced_arrays(X_B_tr, y_B_tr, 2)

y_A_tr_cat = to_categorical(y_A_tr_bal, 2)
y_B_tr_cat = to_categorical(y_B_tr_bal, 2)
y_A_val_cat = to_categorical(y_A_val, 2)
y_B_val_cat = to_categorical(y_B_val, 2)

# ------------------------------
# Building blocks
# ------------------------------
def ConvBNAct(x, f, k=3, sep=False, wd=1e-4, act="relu", dilation=1):
    if sep:
        x = SeparableConv2D(
            f, k, padding="same", use_bias=False, dilation_rate=dilation,
            depthwise_regularizer=l2(wd),
            pointwise_regularizer=l2(wd)
        )(x)
    else:
        x = Conv2D(f, k, padding="same", use_bias=False, dilation_rate=dilation,
                   kernel_regularizer=l2(wd))(x)
    x = BatchNormalization()(x)
    return Activation(act)(x)

def SE(x, r=8):
    c = x.shape[-1]
    s = GlobalAveragePooling2D()(x)
    s = Dense(max(int(c)//r, 8), activation="relu")(s)
    s = Dense(int(c), activation="sigmoid")(s)
    s = Reshape((1,1,int(c)))(s)
    return Multiply()([x, s])

# ------------------------------
# Model A (deeper baseline; ReLU; mostly separable after block1)
# ------------------------------
def build_model_A():
    inp = Input(shape=(28, 28, 1))
    aug = tf.keras.Sequential([
        RandomFlip("horizontal"),
        RandomRotation(0.12),
        RandomZoom(0.10),
        RandomContrast(0.10),
        RandomTranslation(0.05, 0.05),
    ], name="aug_A")
    x = aug(inp)

    # Block 1
    x = ConvBNAct(x, 32, k=3, sep=False, act="relu"); x = ConvBNAct(x, 32, k=3, sep=False, act="relu")
    x = MaxPooling2D()(x); x = Dropout(0.12)(x)

    # Block 2
    x = ConvBNAct(x, 64, k=3, sep=True, act="relu"); x = ConvBNAct(x, 64, k=3, sep=True, act="relu")
    x = MaxPooling2D()(x); x = Dropout(0.20)(x)

    # Block 3
    x = ConvBNAct(x, 96, k=3, sep=True, act="relu"); x = ConvBNAct(x, 96, k=3, sep=True, act="relu")
    x = MaxPooling2D()(x); x = Dropout(0.25)(x)

    # Block 4 (new)
    x = ConvBNAct(x, 128, k=3, sep=True, act="relu"); x = ConvBNAct(x, 128, k=3, sep=True, act="relu")
    x = MaxPooling2D()(x); x = Dropout(0.30)(x)

    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation="relu", kernel_regularizer=l2(1e-4))(x); x = Dropout(0.40)(x)
    feats = Dense(64, activation="relu", name="penultimate")(x)
    logits = Dense(2, name="final_dense")(feats)

    soft = tf.keras.layers.Softmax(name="softmax_head")(logits)
    cheb = ChebyshevSoftmax(CHEB_COEFFS, tau=CHEB_TAU, name="cheb_head")(logits)
    return Model(inp, [soft, cheb]), Model(inp, feats)

# ------------------------------
# Model B (heterogeneous; ELU, wider k early, some dilation + SE attention)
# ------------------------------
def build_model_B():
    inp = Input(shape=(28, 28, 1))
    aug = tf.keras.Sequential([
        RandomFlip("horizontal"),
        RandomRotation(0.14),
        RandomZoom(0.08),
        RandomContrast(0.12),
        RandomTranslation(0.06, 0.06),
    ], name="aug_B")
    x = aug(inp)

    # Block 1
    x = ConvBNAct(x, 48, k=5, sep=False, act="elu"); x = ConvBNAct(x, 48, k=3, sep=True, act="elu")
    x = MaxPooling2D()(x); x = Dropout(0.12)(x)

    # Block 2
    x = ConvBNAct(x, 80, k=3, sep=True, act="elu", dilation=2); x = ConvBNAct(x, 80, k=3, sep=True, act="elu")
    x = MaxPooling2D()(x); x = Dropout(0.20)(x)

    # Block 3 (+SE)
    x = ConvBNAct(x, 128, k=3, sep=False, act="elu"); x = ConvBNAct(x, 128, k=3, sep=True, act="elu", dilation=2)
    x = SE(x, r=8)
    x = MaxPooling2D()(x); x = Dropout(0.28)(x)

    # Block 4
    x = ConvBNAct(x, 160, k=3, sep=True, act="elu"); x = ConvBNAct(x, 160, k=3, sep=True, act="elu")
    x = MaxPooling2D()(x); x = Dropout(0.32)(x)

    x = GlobalAveragePooling2D()(x)
    x = Dense(192, activation="elu", kernel_regularizer=l2(1e-4))(x); x = Dropout(0.35)(x)
    feats = Dense(64, activation="elu", name="penultimate")(x)   # keep 64-d penultimate for transfer demo
    logits = Dense(2, name="final_dense")(feats)

    soft = tf.keras.layers.Softmax(name="softmax_head")(logits)
    cheb = ChebyshevSoftmax(CHEB_COEFFS, tau=CHEB_TAU, name="cheb_head")(logits)
    return Model(inp, [soft, cheb]), Model(inp, feats)

model_A, feat_A = build_model_A()
model_B, feat_B = build_model_B()

# ------------------------------
# AUROC metric for positive class
# ------------------------------
class AUCPos(tf.keras.metrics.Metric):
    def __init__(self, name="auc_pos", **kwargs):
        super().__init__(name=name, **kwargs)
        self._auc = tf.keras.metrics.AUC(curve="ROC", num_thresholds=4095)
    def update_state(self, y_true, y_pred, sample_weight=None):
        return self._auc.update_state(y_true[..., 1], y_pred[..., 1], sample_weight=sample_weight)
    def result(self): return self._auc.result()
    def reset_state(self): self._auc.reset_state()

# ------------------------------
# Losses & metrics (Cheb head uses focal CE)
# ------------------------------
@tf.function
def focal_categorical_crossentropy(y_true, y_pred, gamma=2.0, alpha=0.25):
    y_pred = tf.clip_by_value(y_pred, 1e-6, 1.0 - 1e-6)
    ce = tf.keras.losses.categorical_crossentropy(y_true, y_pred)
    pt = tf.reduce_sum(y_true * y_pred, axis=-1)
    loss = (alpha * tf.pow(1.0 - pt, gamma)) * ce
    return loss

losses = {
    "softmax_head": tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.02),
    "cheb_head":    focal_categorical_crossentropy,
}
loss_weights = {"softmax_head": 0.8, "cheb_head": 1.0}
metrics = {"softmax_head": ["accuracy", AUCPos(name="auc_soft")],
           "cheb_head":    [AUCPos(name="auc_cheb")]}

optA = tf.keras.optimizers.Adam(8e-4)   # slightly lower LR for deeper nets
optB = tf.keras.optimizers.Adam(8e-4)

model_A.compile(optimizer=optA, loss=losses, loss_weights=loss_weights, metrics=metrics, run_eagerly=True)
model_B.compile(optimizer=optB, loss=losses, loss_weights=loss_weights, metrics=metrics, run_eagerly=True)

# ------------------------------
# Callbacks (monitor Cheb AUROC — deploy head)
# ------------------------------
cbs_A = [
    ReduceLROnPlateau(monitor="val_cheb_head_auc_cheb", factor=0.5, patience=5, mode="max", min_lr=2e-5, verbose=1),
    EarlyStopping(monitor="val_cheb_head_auc_cheb", patience=10, mode="max", restore_best_weights=True, verbose=1),
]
cbs_B = [
    ReduceLROnPlateau(monitor="val_cheb_head_auc_cheb", factor=0.5, patience=5, mode="max", min_lr=2e-5, verbose=1),
    EarlyStopping(monitor="val_cheb_head_auc_cheb", patience=10, mode="max", restore_best_weights=True, verbose=1),
]

# Train on OVERSAMPLED arrays (balanced)
yA_targets = [y_A_tr_cat, y_A_tr_cat]
yB_targets = [y_B_tr_cat, y_B_tr_cat]

hist_A = model_A.fit(
    X_A_tr_bal, yA_targets, epochs=60, batch_size=64,
    validation_data=(X_A_val, [y_A_val_cat, y_A_val_cat]),
    callbacks=cbs_A, verbose=1
)
hist_B = model_B.fit(
    X_B_tr_bal, yB_targets, epochs=60, batch_size=64,
    validation_data=(X_B_val, [y_B_val_cat, y_B_val_cat]),
    callbacks=cbs_B, verbose=1
)

# ------------------------------
# Threshold search (use Youden’s J)
# ------------------------------
def best_threshold_youden(y_true, scores):
    # Search more finely around percentiles for stability
    grid = np.unique(np.concatenate([
        np.linspace(0, 1, 2001),
        np.percentile(scores, np.linspace(0, 100, 101))
    ]))
    best_t, best_J = 0.5, -2.0
    for t in grid:
        y_hat = (scores >= t).astype(int)
        tn, fp, fn, tp = confusion_matrix(y_true, y_hat, labels=[0,1]).ravel()
        tpr = tp / (tp + fn + 1e-9)
        tnr = tn / (tn + fp + 1e-9)
        J   = tpr + tnr - 1.0
        if J > best_J:
            best_t, best_J = t, J
    return float(best_t), float(best_J)

def summarize_binary(y_true_int, prob_mat, name="", threshold=None):
    y_scores = prob_mat[:, 1]
    if threshold is None:
        threshold = 0.5
    y_pred = (y_scores >= threshold).astype(int)

    acc = accuracy_score(y_true_int, y_pred)
    try:
        auc_score = roc_auc_score(y_true_int, y_scores)
    except Exception:
        auc_score = np.nan

    prf_micro    = precision_recall_fscore_support(y_true_int, y_pred, average="micro",    zero_division=0)
    prf_macro    = precision_recall_fscore_support(y_true_int, y_pred, average="macro",    zero_division=0)
    prf_weighted = precision_recall_fscore_support(y_true_int, y_pred, average="weighted", zero_division=0)

    print(f"\n===== {name} =====")
    print(f"Threshold used: {threshold:.3f}")
    print(f"Accuracy: {acc:.4f} | AUROC: {auc_score:.4f}")
    print(f"Precision (micro/macro/weighted): {prf_micro[0]:.4f} / {prf_macro[0]:.4f} / {prf_weighted[0]:.4f}")
    print(f"Recall    (micro/macro/weighted): {prf_micro[1]:.4f} / {prf_macro[1]:{'.4f'}} / {prf_weighted[1]:.4f}")
    print(f"F1        (micro/macro/weighted): {prf_micro[2]:.4f} / {prf_macro[2]:.4f} / {prf_weighted[2]:.4f}")

    print("\nClassification report:")
    print(classification_report(y_true_int, y_pred, digits=3, zero_division=0))

    cm = confusion_matrix(y_true_int, y_pred)
    plt.figure(figsize=(5,4))
    plt.imshow(cm, interpolation="nearest")
    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted"); plt.ylabel("True")
    plt.colorbar(); plt.tight_layout(); plt.show()

    return {"accuracy": acc, "auc": auc_score, "cm": cm, "threshold": threshold}

def predict_heads(model, X, batch=256):
    soft, cheb = model.predict(X, batch_size=batch, verbose=0)
    return soft, cheb

# ------------------------------
# Per-site validation (Cheb head is what we deploy)
# ------------------------------
S_A_val, C_A_val = predict_heads(model_A, X_A_val)
S_B_val, C_B_val = predict_heads(model_B, X_B_val)

tA_cheb, _ = best_threshold_youden(y_A_val, C_A_val[:, 1])
tB_cheb, _ = best_threshold_youden(y_B_val, C_B_val[:, 1])

_ = summarize_binary(y_A_val, C_A_val, name=f"Institution A (Validation, Cheb head, J-opt t={tA_cheb:.3f})", threshold=tA_cheb)
_ = summarize_binary(y_B_val, C_B_val, name=f"Institution B (Validation, Cheb head, J-opt t={tB_cheb:.3f})", threshold=tB_cheb)

# Also show Soft head with its OWN Youden-opt thresholds
tA_soft, _ = best_threshold_youden(y_A_val, S_A_val[:, 1])
tB_soft, _ = best_threshold_youden(y_B_val, S_B_val[:, 1])
_ = summarize_binary(y_A_val, S_A_val, name=f"Institution A (Validation, Soft head, J-opt t={tA_soft:.3f})", threshold=tA_soft)
_ = summarize_binary(y_B_val, S_B_val, name=f"Institution B (Validation, Soft head, J-opt t={tB_soft:.3f})", threshold=tB_soft)

# ------------------------------
# B→A Transfer + encrypted calibration demo (binary)
# ------------------------------
# Use A's dense (shared across heads)
W_A, b_A = model_A.get_layer("final_dense").get_weights()

def transfer_probs(extractor_B, W_head, b_head, X, batch_size=256):
    feats = extractor_B.predict(X, batch_size=batch_size, verbose=0)
    probs = np.zeros((feats.shape[0], b_head.shape[0]), dtype=np.float64)
    for i in range(feats.shape[0]):
        logits = np.dot(feats[i], W_head) + b_head
        probs[i] = chebyshev_poly_softmax(logits, CHEB_COEFFS, CHEB_TAU)
    return probs

P_transfer_pre = transfer_probs(feat_B, W_A, b_A, X_test)
_ = summarize_binary(y_test, P_transfer_pre, name="B→A Transfer (Pre-Calibration, Cheb)", threshold=0.5)

# Pick a misclassified test sample under A’s J-opt threshold
y_pred_pre = (P_transfer_pre[:, 1] >= tA_cheb).astype(int)
mis_idx = int(np.where(y_pred_pre != y_test)[0][0]) if np.any(y_pred_pre != y_test) else 0
true_label = int(y_test[mis_idx])

feat_mis_B = feat_B.predict(X_test[mis_idx:mis_idx+1], verbose=0)[0]

# 🔐 Proper TenSEAL flow: build context, then encrypt a 1-D vector
if TENSEAL_AVAILABLE:
    context = ts.context(ts.SCHEME_TYPE.CKKS, 8192, [60, 40, 40, 60])
    context.global_scale = 2**40
    context.generate_galois_keys()
    enc_feat = ts.ckks_vector(context, feat_mis_B.tolist())
else:
    enc_feat = feat_mis_B

# Work on a COPY for the demo (do not alter global head)
scores_before = poly_predict(feat_mis_B, W_A, b_A, CHEB_COEFFS, CHEB_TAU)
print(f"\n🔎 Misclassified Sample (Pre-Cal): index={mis_idx}, True={true_label}, Pred={(scores_before[1] >= tA_cheb):d}")
for c, s in enumerate(scores_before): print(f"Class {c}: {s:.4f}")

W_copy = W_A.copy(); b_copy = b_A.copy()
true_onehot = np.zeros_like(b_copy, dtype=np.float64); true_onehot[true_label] = 1.0
lr = 0.05
loss_history = []
for step in range(20):
    current_pred = encrypted_predict(enc_feat, W_copy, b_copy, CHEB_COEFFS, CHEB_TAU)
    err = true_onehot - np.array(current_pred)
    loss_history.append(float(np.sum(err**2)))
    for j in range(W_copy.shape[1]):
        W_copy[:, j] += lr * err[j] * feat_mis_B
        b_copy[j]    += lr * err[j]

scores_after = poly_predict(feat_mis_B, W_copy, b_copy, CHEB_COEFFS, CHEB_TAU)
print(f"\n✅ Misclassified Sample (Post-Cal demo): index={mis_idx}, True={true_label}, Pred={(scores_after[1] >= tA_cheb):d}")
for c, s in enumerate(scores_after): print(f"Class {c}: {s:.4f}")

plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.bar(np.arange(2)-0.2, scores_before, width=0.4, label="Before")
plt.bar(np.arange(2)+0.2, scores_after,  width=0.4, label="After")
plt.xticks(np.arange(2)); plt.title("Class Probabilities (Misclassified Sample)"); plt.legend()
plt.subplot(1,2,2)
plt.plot(loss_history, marker='o'); plt.title("Calibration Loss Curve"); plt.xlabel("Step"); plt.ylabel("L2 Loss")
plt.tight_layout(); plt.show()

print("\nNote: Models A & B are deeper and differ in kernels/activations/dilation/SE attention.")
print("Chebyshev head remains HE-friendly; penultimate dim kept at 64 for transfer compatibility.")
```
